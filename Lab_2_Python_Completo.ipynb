*Ejercicio 3 - Map*

a. Dada una lista de temperaturas en Celsius, convertirlas a Fahrenheit.

```python
celsius = [0, 20, 37, 100]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)
```

b. Dadas dos listas con la misma cantidad de elementos, generar una nueva sumando cada elementos (primer elemento de lista 1 + primer elemento lista 2).

```python
lista1 = [1, 2, 3, 4]
lista2 = [10, 20, 30, 40]
suma_listas = list(map(lambda x, y: x + y, lista1, lista2))
print(suma_listas)
```

c. Dada una lista de radios de circulos, generar la lista de sus areas.

```python
import math

radios = [1, 2.5, 5, 10]
areas = list(map(lambda r: math.pi * (r ** 2), radios))
print(areas)
```

d. Dada un diccionario, generar una lista de tuplas (nombre, valor+10%)

```python
precios = {"pan": 1000, "leche": 2500, "café": 5000}
resultado = list(map(lambda item: (item[0], item[1] * 1.10), precios.items()))
print(resultado)
```

e. Dada una matriz, multiplicar todos los elementos por 10.

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
matriz_x10 = list(map(lambda fila: list(map(lambda x: x * 10, fila)), matriz))
print(matriz_x10)
```

a. Dada una lista de palabras, filtrar solo aquellas con mas de 4 letras.

c. Dada una lista, hallar el mayor de todos, define una función auxiliar que dado dos numeros retorne el mayor.


```python
from functools import reduce

def mayor(a, b):
    return a if a > b else b

numeros = [12, 45, 7, 89, 23, 56]
resultado = reduce(mayor, numeros)
print(resultado)
```

f. Convertir una palabra a un diccionario que cuente la aparicion de cada letra, ordenar dicho diccionario según la frecuencia de aparicion.


```python
palabra = "programacion"
conteo = {letra: palabra.count(letra) for letra in palabra}
resultado = dict(sorted(conteo.items(), key=lambda item: item[1], reverse=True))
print(resultado)
```

*Ejercicio 6 - Minimax*


Juego NIM usando el algoritmo Minimax (MAX = Computadora, MIN = Humano, primer turno MIN).


```python
def minimax(piedras, es_max):
    if piedras == 0:
        return -1 if es_max else 1

    movimientos = [m for m in [1, 2, 3] if m <= piedras]

    if es_max:
        mejor_valor = -float("inf")
        for m in movimientos:
            valor = minimax(piedras - m, False)
            mejor_valor = max(mejor_valor, valor)
        return mejor_valor
    else:
        mejor_valor = float("inf")
        for m in movimientos:
            valor = minimax(piedras - m, True)
            mejor_valor = min(mejor_valor, valor)
        return mejor_valor

def mejor_jugada_cpu(piedras):
    mejor_valor = -float("inf")
    jugada = 1
    for m in [1, 2, 3]:
        if m <= piedras:
            valor = minimax(piedras - m, False)
            if valor > mejor_valor:
                mejor_valor = valor
                jugada = m
    return jugada

def jugar_nim(piedras=10):
    turno_humano = True
    while piedras > 0:
        print(f"\nPiedras restantes: {piedras}")
        if turno_humano:
            jugada = int(input("Tu turno (quitar 1, 2 o 3 piedras): "))
            while jugada not in [1, 2, 3] or jugada > piedras:
                jugada = int(input("Movimiento inválido. Elige 1, 2 o 3 (sin pasarte de las piedras): "))
            piedras -= jugada
            if piedras == 0:
                print("¡Tomaste la última piedra! Ganaste (MIN).")
                return
        else:
            jugada = mejor_jugada_cpu(piedras)
            print(f"La CPU (MAX) quita {jugada} piedra(s).")
            piedras -= jugada
            if piedras == 0:
                print("La CPU tomó la última piedra. ¡Gana la CPU (MAX)!")
                return
        turno_humano = not turno_humano

jugar_nim(10)
```
